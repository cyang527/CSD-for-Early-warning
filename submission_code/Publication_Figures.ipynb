# Publication figures
Generate the main-text and supplementary figures. The study schematic is supplied as artwork. Figure data and images are written under results/Publication_Figures.

```python
from pathlib import Path
import sys

cwd = Path.cwd().resolve()
code_dir = next(
    (candidate for folder in (cwd, *cwd.parents)
     for candidate in (folder, folder / "submission_code")
     if (candidate / "manuscript_replay.py").is_file()
     and (candidate.parent / "submission_inputs").is_dir()),
    None,
)
if code_dir is None:
    raise FileNotFoundError("Open the project containing submission_code and submission_inputs.")
sys.path.insert(0, str(code_dir))
```

```python
from IPython.display import Image, display
from plot_publication_figures import build_main_figures, build_si_figures
from release_replay import RESULTS, collect

outcomes, histories = collect()
output_dir = RESULTS / "Publication_Figures"
```

```python
main_figures = build_main_figures(output_dir / "main", outcomes, histories)
for figure in main_figures:
    display(Image(filename=str(figure.with_suffix(".png")), width=790))
```

```python
si_figures = build_si_figures(output_dir / "si", outcomes, histories)
[figure.relative_to(RESULTS).as_posix() for figure in si_figures]
```
